```python
def plot_item_price_trend(df, item_name):
    temp = df.copy()
    temp["date"] = pd.to_datetime(temp["date"])

    item_df = temp[temp["item"]==item_name]
    item_df = item_df.sort_values("date")

    plt.figure(figsize = (12,6))
    plt.plot(item_df["date"], item_df["price"])

    plt.title(f"{item_name} Prices Over Time")
    plt.xlabel("Date")
    plt.ylabel("Price")
    plt.xticks(rotation = 45)
    plt.grid(True)
    plt.show()
```

```python
plot_item_price_trend(market_prices, "Pesticide")
```

### weekly alert triage analysis

```python
sensor = datasets["daily_sensor_readings"].copy()

sensor["date"] = pd.to_datetime(sensor["date"])
sensor["week"] = sensor["date"].dt.isocalendar().week
sensor["year"] = sensor["date"].dt.year
```

```python
sensor["plant_stress_index"].describe()
sensor["date"]
```

```python
sensor["high_stress_flag"] = (sensor["plant_stress_index"] >= 0.70).astype(int)
```

```python
weekly_triage = (
    sensor
    .groupby(["year", "week", "farm_id", "greenhouse_id", "plot_id", "crop"])
    .agg(
        avg_stress=("plant_stress_index", "mean"),
        max_stress=("plant_stress_index", "max"),
        high_stress_days=("high_stress_flag", "sum"),
        alert_days=("alert_flag", "sum"),
        action_days=("action_taken", "sum"),
        avg_action_delay=("action_delay_days", "mean"),
        avg_post_action_stress_delta=("post_action_stress_delta_3d", "mean")
    )
    .reset_index()
)

weekly_triage["alert_response_rate"] = (
    weekly_triage["action_days"] / weekly_triage["alert_days"]
).replace([np.inf, -np.inf], np.nan).fillna(0)
```

```python
weekly_triage["triage_score"] = (
    weekly_triage["high_stress_days"] * 3
    + weekly_triage["alert_days"] * 2
    + weekly_triage["avg_stress"] * 5
    + weekly_triage["avg_action_delay"].fillna(0)
    - weekly_triage["alert_response_rate"] * 2
)
```

```python
weekly_triage = weekly_triage.sort_values(
    ["year", "week", "triage_score"],
    ascending=[True, True, False]
)

weekly_triage.head(20)
```

### Seasonal precision ROI Analysis

```python
plot_sensor_summary = (
    sensor
    .groupby("plot_id")
    .agg(
        avg_stress=("plant_stress_index", "mean"),
        max_stress=("plant_stress_index", "max"),
        high_stress_days=("high_stress_flag", "sum"),
        total_alerts=("alert_flag", "sum"),
        total_actions_taken=("action_taken", "sum"),
        avg_action_delay=("action_delay_days", "mean"),
        avg_post_action_stress_delta=("post_action_stress_delta_3d", "mean")
    )
    .reset_index()
)

plot_sensor_summary["alert_response_rate"] = (
    plot_sensor_summary["total_actions_taken"] / plot_sensor_summary["total_alerts"]
).replace([np.inf, -np.inf], np.nan).fillna(0)
```

```python
costs = datasets["daily_input_costs"].copy()
costs["date"] = pd.to_datetime(costs["date"])

plot_cost_summary = (
    costs
    .groupby("plot_id")
    .agg(
        daily_input_cost_total=("daily_total_input_cost", "sum"),
        precision_cost_total=("daily_precision_cost", "sum"),
        routine_cost_total=("daily_routine_cost", "sum"),
        precision_actions_total=("daily_precision_actions_count", "sum"),
        total_actions_count=("daily_total_actions_count", "sum")
    )
    .reset_index()
)

plot_cost_summary["precision_cost_share"] = (
    plot_cost_summary["precision_cost_total"] / plot_cost_summary["daily_input_cost_total"]
).replace([np.inf, -np.inf], np.nan).fillna(0)
```

```python
season = datasets["season_summary"].copy()
plots = datasets["plots"].copy()
greenhouses = datasets["greenhouses"].copy()
farms = datasets["farms"].copy()
```

```python
plot_analysis = (
    plots
    .merge(plot_sensor_summary, on="plot_id", how="left")
    .merge(plot_cost_summary, on="plot_id", how="left")
    .merge(season, on="plot_id", how="left")
    .merge(
        greenhouses,
        on=["greenhouse_id", "farm_id"],
        how="left",
        suffixes=("", "_greenhouse")
    )
    .merge(farms, on="farm_id", how="left")
)

plot_analysis.head()
```

### Business KPIs

```python
plot_analysis["profit_per_m2"] = (
    plot_analysis["season_profit_cad"] / plot_analysis["plot_area_m2"]
)

plot_analysis["revenue_per_m2"] = (
    plot_analysis["season_revenue_cad"] / plot_analysis["plot_area_m2"]
)

plot_analysis["cost_per_m2"] = (
    plot_analysis["total_cost_cad"] / plot_analysis["plot_area_m2"]
)

plot_analysis["marketable_yield_kg_m2"] = (
    plot_analysis["season_yield_kg_m2"] * plot_analysis["marketable_ratio"]
)

plot_analysis["precision_benefit_per_m2"] = (
    plot_analysis["precision_benefit_cad"] / plot_analysis["plot_area_m2"]
)
```

```python
plot_analysis["roi_segment"] = pd.qcut(
    plot_analysis["season_roi"],
    q=4,
    labels=["Low ROI", "Medium-Low ROI", "Medium-High ROI", "High ROI"]
)

plot_analysis["stress_segment"] = pd.qcut(
    plot_analysis["avg_stress"],
    q=4,
    labels=["Low Stress", "Medium-Low Stress", "Medium-High Stress", "High Stress"]
)
```

### Compare precision actions vs outcomes

```python
precision_summary = (
    plot_analysis
    .groupby("treatment")
    .agg(
        plots=("plot_id", "count"),
        avg_stress=("avg_stress", "mean"),
        avg_alerts=("total_alerts", "mean"),
        avg_response_rate=("alert_response_rate", "mean"),
        avg_precision_cost=("precision_cost_total", "mean"),
        avg_precision_cost_share=("precision_cost_share", "mean"),
        avg_yield=("season_yield_kg_m2", "mean"),
        avg_marketable_ratio=("marketable_ratio", "mean"),
        avg_profit=("season_profit_cad", "mean"),
        avg_roi=("season_roi", "mean"),
        avg_precision_benefit=("precision_benefit_cad", "mean")
    )
    .reset_index()
    .sort_values("avg_roi", ascending=False)
)

precision_summary
```

### Analyze treatement effectiveness against control

```python
control_metrics = (
    plot_analysis[plot_analysis["treatment"] == "Control"]
    .agg(
        control_yield=("season_yield_kg_m2", "mean"),
        control_profit=("season_profit_cad", "mean"),
        control_roi=("season_roi", "mean"),
        control_stress=("avg_stress", "mean")
    )
)

control_yield = plot_analysis.loc[
    plot_analysis["treatment"] == "Control", 
    "season_yield_kg_m2"
].mean()

control_profit = plot_analysis.loc[
    plot_analysis["treatment"] == "Control", 
    "season_profit_cad"
].mean()

control_roi = plot_analysis.loc[
    plot_analysis["treatment"] == "Control", 
    "season_roi"
].mean()

control_stress = plot_analysis.loc[
    plot_analysis["treatment"] == "Control", 
    "avg_stress"
].mean()
```

```python
treatment_effect = (
    plot_analysis
    .groupby("treatment")
    .agg(
        avg_yield=("season_yield_kg_m2", "mean"),
        avg_profit=("season_profit_cad", "mean"),
        avg_roi=("season_roi", "mean"),
        avg_stress=("avg_stress", "mean"),
        avg_precision_benefit=("precision_benefit_cad", "mean")
    )
    .reset_index()
)

treatment_effect["yield_vs_control"] = (
    treatment_effect["avg_yield"] - control_yield
)

treatment_effect["profit_vs_control"] = (
    treatment_effect["avg_profit"] - control_profit
)

treatment_effect["roi_vs_control"] = (
    treatment_effect["avg_roi"] - control_roi
)

treatment_effect["stress_vs_control"] = (
    treatment_effect["avg_stress"] - control_stress
)

treatment_effect.sort_values("profit_vs_control", ascending=False)
```
